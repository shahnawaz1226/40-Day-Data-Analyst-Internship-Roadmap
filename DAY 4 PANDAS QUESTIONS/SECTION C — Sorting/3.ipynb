```python
import pandas as pd

data = {
    "Name": ["Ali", "Sara", "John", "Emma"],
    "Marks": [78, 92, 65, 88]
}

df = pd.DataFrame(data)
```

```python
# Q6
# Sort students by marks ascending.

df.sort_values('Marks', ascending=False)
```

```python
# Q7
# Sort descending.

df.sort_values('Marks')
```
