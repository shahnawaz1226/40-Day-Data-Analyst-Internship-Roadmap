```python
# Q4
# Use .agg() to calculate:

# mean
# max
# min
# sum
```

```python
import pandas as pd
data = {
    "Department": ["HR", "IT", "IT", "Sales", "HR"],
    "Salary": [30000, 50000, 60000, 45000, 35000]
}

df = pd.DataFrame(data)

df.groupby('Department')['Salary'].agg(['min', 'max', 'sum', 'mean'])
```

```python
# Q5
# Count employees department-wise.
df['Department'].value_counts()
```
