```python
import pandas as pd
```

```python
df = pd.read_csv('expense.csv')
df
```

```python
Total_expense = df['amount'].sum()
print("Total expense:", Total_expense)
```

```python
expense_category = df.groupby(by='category')['amount'].sum()
expense_category
```

```python
expense_category.idxmax()
```

```python
date_wise = df.groupby(by='date')['amount'].sum()
date_wise
```
