CSV: marks.csv

Columns: name, subject, marks

Subject-wise average

Overall class average

Fail students (marks < 33)

```python
import pandas as pd
```

```python
df = pd.read_csv('marks.csv')
df
```

```python
subject_avg = df.groupby("subject")["marks"].mean()
print("\nSubject-wise Average Marks:")
print(subject_avg)
```

```python
class_avrg = df["marks"].mean()
print("Average of class:", class_avrg)
```

```python
failed = df[df['marks'] < 33]
print(failed)
```
