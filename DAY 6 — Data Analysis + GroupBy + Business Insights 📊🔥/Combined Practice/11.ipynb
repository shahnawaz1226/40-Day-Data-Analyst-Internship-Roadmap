```python
import pandas as pd
```

```python
df = pd.read_csv('student_dataset.csv')
df
```

```python
df.insert(5, 'Total_marks', df['Math'] + df['Physics'] + df['Chemistry'])
df
```

```python
df.insert(6, 'Percentage', (df['Total_marks']/300)*100)
df
```

```python
def grade(a):
    if a>=90:
        return "A+"
    elif a>=75:
        return "A"
    elif a>=60:
        return "B"
    elif a>=40:
        return "C"
    else:
        return "Fail"
```

```python
df.insert(8, "Grade1", df['Percentage'].apply(grade))
df
```

```python
topper = df.loc[df['Percentage'].idxmax(), 'Student_Names']
topper
```

```python
physics_avg = df['Physics'].mean()
print("Physics Average:", physics_avg)

math_avg = df['Math'].mean()
print("Math Average:", math_avg)

chemistry_avg = df['Chemistry'].mean()
print("Chemistry Average:", round(chemistry_avg, 2))
```
