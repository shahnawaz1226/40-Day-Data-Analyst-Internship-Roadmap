```python
import pandas as pd
import numpy as np
```

```python
data = pd.read_csv('attendance.csv')
data
```

```python
less_attendance = data[data['attendance_percentage']<75]
less_attendance
```

```python
mean_att = data['attendance_percentage'].mean()
print('Attendance Average:', mean_att)

att_mean = np.array(data['attendance_percentage'])
np.mean(att_mean)
```

```python
top3 = data.sort_values(by="attendance_percentage", ascending=False).head(3)
print("Top 3 Attendance:\n", top3)
```
