```python
def bucket(x):
    if x < 40000:
        return "Low"
    elif 40000 <= x <= 80000:
        return "Medium"
    elif x > 80000:
        return "High"
    else:
        return "Missing value"
```

```python
df['salary_bucket'] = df['salary'].apply(bucket)
df
```
