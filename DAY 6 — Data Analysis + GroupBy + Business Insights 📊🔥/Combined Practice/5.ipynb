Random numbers generate karo NumPy se (10 numbers)

DataFrame banao

Even numbers filter karo

Mean, std calculate karo

```python
import numpy as np
import pandas as pd
```

```python
random_numbers = np.random.randint(1, 100, 10)
```

```python
df = pd.DataFrame({
    "Numbers": random_numbers
})
```

```python
evens = df[df['Numbers']%2==0]
evens
```

```python
mean = df['Numbers'].mean()
print(mean)
mean_ = np.mean(random_numbers)
print(mean_)

std = df['Numbers'].std()
print(std)
std_ = np.std(random_numbers)
print(std_)
```
