🧠 Pandas Questions — Social Media Dataset
🔹 Intermediate

Clean missing usernames.

Standardize date format.

Replace NaN likes/comments with median.

Convert verified to boolean.

Compute engagement rate = (likes + comments + shares) / followers.

```python
import pandas as pd
```

```python
df = pd.read_csv('social_media_messy.csv')
df
```

```python
df['user'] = df['user'].fillna('unknown_user')
```

```python
df['posted_at'] = pd.to_datetime(df['posted_at'], errors='coerce', format='mixed')
```

```python
df['likes'] = df['likes'].fillna(df['likes'].median())
df['comments'] = df['comments'].fillna(df['comments'].median())
```

```python
def verified(x):
    if x=='Yes':
        return True
    else:
        return False
```

```python
df['verified'] = df['verified'].apply(verified)
```

```python
df['engagement_rate'] = ((df['likes'] + df['comments'] + df['shares'])/df['followers']).round(3)
```
