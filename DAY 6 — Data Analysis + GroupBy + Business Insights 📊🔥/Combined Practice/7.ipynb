Weather dataset:

Columns: day, temperature

Max temp

Min temp

Average temp using NumPy

DataFrame me new column add karo: "Status"

temp > avg → "Hot"

else → "Cool"

```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('weather.csv')
df
```

```python
max_temp = df.loc[[df['temperature'].idxmax()]]
min_temp = df.loc[[df['temperature'].idxmin()]]
```

```python
max_temp
```

```python
min_temp
```

```python
avg_temp = np.mean(df['temperature'])
print("Average Temperature:", avg_temp)
```

```python
df["Status"] = df["temperature"].apply(lambda x: "Hot" if x > avg_temp else "Cool")
df
```
