sales.csv (date, product, amount)
Tasks:

Total sales

Product-wise total sales

Highest selling product

```python
df = pd.read_csv('sales.csv')
df
```

```python
total_sales = df['amount'].sum()
print("₹", total_sales)
```

```python
product_sales = df.groupby(by='product')['amount'].sum()
product_sales
```

```python
prdct_sale = pd.DataFrame(product_sales)
prdct_sale
```

```python
prdct_sale.loc[prdct_sale['amount'].idxmax()]
```
