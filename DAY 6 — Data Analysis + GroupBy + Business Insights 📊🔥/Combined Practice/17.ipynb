Random 20 numbers lo → Z-score normalization using NumPy

Z-score formula:

𝑧 = 𝑥−𝜇 / 𝜎

where
μ = mean

σ = standard deviation

```python
import numpy as np
import pandas as pd
```

```python
array = np.random.randint(1,100,20)
array
```

```python
mean = np.mean(array)
mean
```

```python
std = np.std(array)
std
```

```python
z = (array - mean)/std
z
```

```python
pd.DataFrame({
    "Original": array,
    "Z-score": z
})
```
