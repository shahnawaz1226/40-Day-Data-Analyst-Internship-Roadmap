Movies dataset:

Columns: name, rating, year

Highest rated movie

Movies after 2020

Average rating using NumPy

```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('movies.csv')
df
```

```python
highest_rated = df.loc[[df['rating'].idxmax()]]
highest_rated
```

```python
after_20 = df[df['year']>2020]
after_20
```

```python
avg_rating = np.mean(df['rating'])
print("Average rating:", avg_rating)
```
