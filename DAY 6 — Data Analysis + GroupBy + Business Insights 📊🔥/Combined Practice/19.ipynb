Create a new column total_amount = price * quantity.

```python
df['total_amount'] = df['price'] * df['quantity']
df
```

```python
total_revenue = df['total_amount'].sum()
total_revenue
```

```python
avg_order_val = df['total_amount'].mean()
avg_order_val
```

```python
# List unique cities and payment modes.
df['city'].unique()
```

```python
df['payment_mode'].unique()
```

```python
# Sort orders by total_amount descending.
df.sort_values(by='total_amount', ascending=False)
```

```python
# Filter all Electronics orders above ₹10,000.
df[(df['category']== 'Electronics') & (df['total_amount']>10000)]
```

```python
df.columns = df.columns.str.lower()
df.columns = df.columns.str.replace(" ", "_")
```

```python
date_time = pd.to_datetime(df['order_date'])
date_time.dt.day
```

```python
date_time.dt.month
```

```python
df["order_date"] = pd.to_datetime(df["order_date"])

df["order_day"] = df["order_date"].dt.day
df["order_month"] = df["order_date"].dt.month
```
