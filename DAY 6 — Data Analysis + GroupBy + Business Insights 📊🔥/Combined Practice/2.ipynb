```python
import numpy as np
import pandas as pd
```

```python
Data = {
    "Name": ['A', 'B', 'C'],
    "Age": [22, 25 , 21],
    "Salary": [25000, 40000, 18000]
}
```

```python
df = pd.DataFrame(Data)
df
```

```python
Salary_array = np.array(df['Salary'])
Salary_array
# print("Array of Salary:", Salary_array)
```

```python
Mean_Salary = df['Salary'].mean()
High_Salaray = df['Salary'].max()
print(Mean_Salary)
print(High_Salaray)
```

```python
df[df['Salary'] > 20000]
```
