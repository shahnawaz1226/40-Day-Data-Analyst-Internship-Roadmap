Task: Attendance ko 3 buckets me baantna

High → > 85

Medium → 70–85

Low → < 70

```python
import pandas as pd
```

```python
df = pd.read_csv('attendance.csv')
df
```

```python
def att_category(x):
    if x>85:
        return "High"
    elif 70 <= x <= 85:
        return "Medium"
    else:
        return "Low"
```

```python

df['Category'] = df['attendance_percentage'].apply(att_category)
df
```
