Columns: name, marks

Max & Min marks
Topper ka naam print karo

```python
import pandas as pd
```

```python
df = pd.read_csv('student_dataset.csv')
df
```

```python
# df['Marks'] = df['Math'] + df['Physics'] + df['Chemistry']
```

```python
df.insert(5, "Marks", df['Math'] + df['Physics'] + df['Chemistry'])
df
```

```python
avg_marks = df['Marks'].mean()
max_mark = df['Marks'].max()
min_marks = df['Marks'].min()
topper = df.loc[df['Marks'].idxmax(), 'Student_Names']
lowest = df.loc[df['Marks'].idxmin(), 'Student_Names']
```

```python
print('Average Marks:', avg_marks)
print('Maximum Marks:', max_mark)
print('Minimum Marks:', min_marks)
print('Topper Name:', topper, max_mark)
print('Lowest Name:', lowest, min_marks)
# print(topper1)
```
