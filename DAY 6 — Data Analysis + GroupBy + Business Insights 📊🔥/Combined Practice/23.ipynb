```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('employees2.csv')
df
```

```python
df.groupby('department')['salary'].mean()
```

```python
df.groupby('department').agg({
    "salary": ["mean","max"]
})
```

```python
df.groupby('department')['salary'].max() 
```

```python
df.sort_values(by="salary", ascending=False).head(10)
```

```python
df[(df['salary']> df['salary'].mean())]
```

```python
df[(df['experience_years']>5) & (df['salary']>900000)]
```

```python
pd.pivot_table(df, values="salary", index="department", aggfunc="mean")
```

```python
null_val = df['salary'].isnull()
null_val.sum()
```

```python
df['salary'].fillna(df.groupby('department')['salary'].transform('mean'), inplace=True)
```
