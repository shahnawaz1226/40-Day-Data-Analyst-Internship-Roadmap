Row-wise mean

```python
import numpy as np
import pandas as pd
```

```python
Matrix = np.random.randint(1, 100, (5,5))
Matrix
```

```python
df = pd.DataFrame(Matrix)
df
```

```python
row_mean = df.mean(axis=1)
print("Row-wise average\n",row_mean, sep="")
```

```python
col_max = df.max()
print("Column-wise mmax\n",col_max, sep="")
```
