```python
# Dataset:
# data = {
#     "City": ["Delhi", "Mumbai", "Delhi", "Chennai", "Mumbai", "Delhi"],
#     "Sales": [2000, 3000, 2500, 1800, 3500, 4000]
# }
# Q6
# Find total sales city-wise.
```

```python
data = {
    "City": ["Delhi", "Mumbai", "Delhi", "Chennai", "Mumbai", "Delhi"],
    "Sales": [2000, 3000, 2500, 1800, 3500, 4000]
}

import pandas as pd
df = pd.DataFrame(data)
```

```python
df.groupby('City')['Sales'].sum()
```

```python
# Q7
# Find average sales city-wise.
df.groupby('City')['Sales'].mean()
```

```python
# Q8
# Find city with highest total sales.
df.groupby('City')['Sales'].sum().sort_values().tail(1)
```

```python
# Q9
# Sort cities by sales descending.
df.groupby('City')['Sales'].sum().sort_values(ascending=False)
```
