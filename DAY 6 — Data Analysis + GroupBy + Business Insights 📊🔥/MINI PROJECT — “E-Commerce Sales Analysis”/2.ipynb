```python
import pandas as pd

data = {
    "Product": ["Laptop", "Phone", "Laptop", "Tablet", "Phone"],
    "Revenue": [50000, 30000, 60000, 20000, 35000]
}
df = pd.DataFrame(data)
```

```python
# total revenue product-wise
df.groupby('Product')['Revenue'].sum()
```

```python
# average revenue
df.groupby('Product')['Revenue'].mean()
```

```python
# highest revenue product
df.groupby('Product')['Revenue'].sum().sort_values(ascending=False).head(1)
```

```python
# sorted revenue ranking
df.groupby('Product')['Revenue'].sum().sort_values(ascending=False)
```
