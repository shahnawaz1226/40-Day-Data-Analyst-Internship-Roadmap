```python
import pandas as pd

data = {
    "Category": ["Electronics", "Clothing", "Electronics", "Food", "Clothing", "Food"],
    "Sales": [5000, 3000, 7000, 2000, 4000, 3500],
    "City": ["Delhi", "Mumbai", "Delhi", "Chennai", "Mumbai", "Delhi"]
}

df = pd.DataFrame(data)
```

```python
# total sales category-wise
df.groupby('Category')['Sales'].agg(['sum', 'mean'])
```

```python
# highest selling category
df.groupby('Category')['Sales'].sum().sort_values().tail(1)
```

```python
# city-wise sales
df.groupby('City')['Sales'].sum()
```

```python
# top performing city
df.groupby('City')['Sales'].sum().sort_values().tail(1)
```

```python
# sort categories by sales
df.groupby('Category')['Sales'].sum().sort_values(ascending=False)
```

```python
# count transactions per category
df['Category'].value_counts()
```
