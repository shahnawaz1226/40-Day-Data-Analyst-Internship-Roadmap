```python
import pandas as pd

# Create DataFrame
df = pd.DataFrame({
    "Product": ["Laptop", "Phone", "Tablet"],
    "Price": [55000, 25000, 18000]
})
```

```python
# add GST column
df['GST'] = df['Price']*0.18
```

```python
# create FinalPrice column
df['FinalPrice'] = df['Price'] + df['GST']
df
```

```python
# filter products above ₹25000
df[df['FinalPrice']>25000]
```

```python
# sort highest to lowest price
df.sort_values('FinalPrice', ascending=False)
```
