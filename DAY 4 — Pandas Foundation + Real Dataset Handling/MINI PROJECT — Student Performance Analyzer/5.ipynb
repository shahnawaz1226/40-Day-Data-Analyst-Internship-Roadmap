```python
import pandas as pd

data = {
    "Name": ["Ali", "Sara", "John", "Emma", "David"],
    "Marks": [78, 92, 65, 88, 55],
    "Attendance": [85, 95, 70, 90, 60]
}
df = pd.DataFrame(data)

# average marks
df['Marks'].mean()
```

```python
# topper name
df.sort_values('Marks', ascending=False).head(1)
```

```python
# lowest scorer
df.sort_values('Marks').head(1)
```

```python
# students above 80
df[df['Marks']>80]
```

```python
# attendance below 75
df[df['Attendance']<75]
```

```python
# create bonus column
df['Bonus'] = 5
```

```python
# final marks after bonus
df['FinalMarks'] = df['Marks']+df['Bonus']
```

```python
# sort by final marks
df.sort_values('FinalMarks', ascending=False)
```
