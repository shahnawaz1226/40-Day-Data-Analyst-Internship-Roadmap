```python
import pandas as pd

data = {
    "Name": ["Ali", "Sara", "Ali", "John"],
    "Marks": [80, 90, 80, 70]
}

df = pd.DataFrame(data)
```

```python
df.duplicated()
```

```python
# Q6
# Remove duplicates.
df = df.drop_duplicates()
```

```python
# Check data types of columns.
print(df.dtypes)
```

```python
# Convert int column to float.
df['Marks'] = df['Marks'].astype(float)
df.dtypes
```
