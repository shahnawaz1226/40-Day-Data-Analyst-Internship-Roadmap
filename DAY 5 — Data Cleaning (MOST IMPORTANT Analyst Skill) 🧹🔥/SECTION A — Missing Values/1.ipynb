```python
import pandas as pd

data = {
    "Name": ["A", "B", None, "D"],
    "Marks": [90, None, 78, 85]
}
df = pd.DataFrame(data)

df.isnull()
```

```python
# Q2
# Count missing values column-wise.
df.isnull().sum()
```

```python
# Q3
# Fill missing marks with average marks.
df['Marks'].fillna(df['Marks'].mean())
```

```python
# Q4
# Fill missing names with “Unknown”.

df['Name'].fillna("Unknown")
```
