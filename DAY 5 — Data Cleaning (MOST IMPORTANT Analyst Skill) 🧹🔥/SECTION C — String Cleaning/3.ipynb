```python
# Dataset:
# cities = [" Delhi ", "MUMBAI", "chennai", " Delhi"]
# Q7
# Convert all to lowercase.
# Q8
# Remove spaces.
# Q9
# Make formatting consistent.

# Expected:
# delhi
# mumbai
# chennai
# delhi
```

```python
cities = [" Delhi ", "MUMBAI", "chennai", " Delhi"]

cleaned_city = []

for city in cities:
    city = city.lower()
    city = city.strip()
    cleaned_city.append(city)

for i in cleaned_city:
    print(i)
```
